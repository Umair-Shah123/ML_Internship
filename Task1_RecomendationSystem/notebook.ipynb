```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("../data/Dataset.csv")
display(df.head())
```

```python
display(df.tail())
```

```python

df['country']=df['country'].replace("Not Given","")
```

```python

df['director']=df['director'].replace("Not Given","")
print(df['director'])
```

```python
df=df.drop_duplicates(subset='title').reset_index(drop=True)
```

```python
print((df["director"] == "Not Given").sum())
```

```python
print(df.shape)
```

```python
def clean_list(text):
    cleaned=[]
    for piece in text.split(","):
        piece=piece.strip().replace("","").lower()
        cleaned.append(piece)
    return "".join(cleaned)
print(clean_list("Raúl Campos, Jan Suter"))    
print(repr(clean_list("")))     
```

```python
df['soup']=(
df['listed_in'].apply(clean_list)+" "+
df['director'].apply(clean_list)+""+
df['country'].apply(clean_list)+""+
df['rating'].str.replace(" "," ").str.lower()+""+
df['type'].str.replace(" "," ").str.lower()


)
df[['title','soup']].head(5)
```

```python
print(df[['title','soup']].sample(5))
print(df['soup'].isnull().sum())
```

Step 2: Convert text data into machine-readable format.

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer()
vectors=cv.fit_transform(df['soup'])
print(vectors.shape)
```

Step 3: Calculate content similarity scores.

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity=cosine_similarity(vectors)
print(similarity.shape)
```

Step 4: Generate recommendations for selected titles

```python
def recommend(title):
    if title not in df['title'].values:
       print ("Movie Not Found")
       return
    title_index=df[df['title']==title].index[0]
    # print("Before Sorting",similarity_score[:5])
    similar_scores=list(enumerate(similarity[title_index]))
    similar_scores=sorted(similar_scores,key=lambda x:x[1],reverse=True)
    # print("After Sorting",similarity_score[:10])
    print(similar_scores[1:11])
    for index,score in similar_scores[1:11]:
      print(df.iloc[index]['title'],score)
```

```python
recommend("Ganglands")
```

```python
recommend("Lupin")
```

```python
recommend("Midnight Mass")
```

Step 5:Evaluation

```python
from sklearn.metrics import confusion_matrix ,classification_report

def evaluate(title,threshold=0.5):
    title_index=df[df['title']==title].index[0]

    scores=similarity[ title_index]
    y_true=[]
    y_pred=[]

    for index,score in enumerate(scores):
        if index==title_index:
            continue
        actual=1 if score>=threshold else 0
        pred=1 if score>=threshold else 0

        y_true.append(actual)
        y_pred.append(pred)

    print("Confusion Matrix:")
    print(confusion_matrix(y_true,y_pred))

    print("\n Classification Report")
    print(classification_report(y_true,y_pred,target_names=["Not Relevant","Relevant"],zero_division=0))
evaluate("Ganglands")    
  
```
