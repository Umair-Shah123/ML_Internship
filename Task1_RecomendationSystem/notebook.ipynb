```python
def clean_list(text):
    cleaned=[]
    for piece in text.split(","):
        piece=piece.strip().replace("","").lower()
        cleaned.append(piece)
    return "".join(cleaned)
print(clean_list("Raúl Campos, Jan Suter"))    # raúlcampos jansuter
print(repr(clean_list("")))     
```

```python
df['soup']=(
df['listed_in'].apply(clean_list)+" "+
df['director'].apply(clean_list)+""+
df['country'].apply(clean_list)+""+
df['rating'].str.replace(" "," ").str.lower()+""+
df['type'].str.replace(" "," ").str.lower()


)
df[['title','soup']].head(5)
```
