```python
missing_values=df.isnull().sum()
print("Before CLeaning")
print(missing_values)
print("\nMissing Percentage")
print(df.isnull().sum()/len(df)*100)
```

```python
df['date_added']=pd.to_datetime(df['date_added'],errors='coerce')

df['added_year']=df['date_added'].dt.year
df['added_month']=df['date_added'].dt.month
df['added_day']=df['date_added'].dt.day
df['added_weekday']=df['date_added'].dt.day_of_week
df[['date_added', 'added_year', 'added_month',
    'added_day', 'added_weekday']].head()
```

```python
print(df['duration'].head())
print("Content Types")
print(df['type'].value_counts())
```

```python
df['duration_value']=df['duration'].str.extract(r'(\d+)')

df['duration_value']=pd.to_numeric(df['duration_value'],errors='coerce')
df['duration_unit']=np.where(df['type']=='Movie','Minutes','Season')

df[['type', 'duration', 'duration_value',
    'duration_unit']].head(5)
```

```python
print(df['listed_in'].head(5))
```

```python
df['genre_count']=df['listed_in'].apply(lambda x:len(x.split(',')))
df[['title','listed_in','genre_count']].head(5)
```

```python
df['primary_genre'] = df['listed_in'].apply(lambda x: x.split(',')[0].strip())
df[['title','listed_in','primary_genre']].head(10)
```

```python
df['country_count'] = df['country'].apply(
    lambda x: len(x) if isinstance(x, list)
    else len(str(x).split(','))
)
df['primary_country'] = df['country'].apply(lambda x: x.split(',')[0].strip())
df[['title','country','country_count','primary_country']].head(10)
```

```python

df['has_director'] = np.where(
    df['director'].str.strip().str.lower() == 'unknown', 0, 1)
df[['title', 'director', 'has_director']].head(10)
```

```python
#i will take refrence age as 2026
df['content_age']=2026-df['release_year']
df[['title','release_year','content_age']].head(10)
```

```python
print("Total Columns:", len(df.columns))

print("\nAll Columns:")
print(df.columns.tolist())
```

```python
print("Original Dataset Shape:", df.shape)
print("\nNew Engineered Features:")
new_features = [
    'added_year',
    'added_month',
    'added_day',
    'added_weekday',
    'duration_value',
    'duration_unit',
    'genre_count',
    'primary_genre',
    'country_count',
    'primary_country',
    'has_director',
    'content_age'
]
print(df[new_features].head())
print("\nFinal Dataset Shape:", df.shape)
```

```python
print("Content Type Distribution:")
print(df['type'].value_counts())
print("Percentage Distribution:")
print(df['type'].value_counts(normalize=True) * 100)
```

```python
y = df['type']
X = df[
    [ 'release_year','added_year', 'added_month','added_day', 'added_weekday','duration_value',
     'genre_count','primary_genre','country_count', 'primary_country', 'has_director','content_age','rating'  ]
]


print("X Shape:", X.shape)
print("y Shape:", y.shape)

print("\nInput Features:")
print(X.columns.tolist())
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split( X, y,test_size=0.20,random_state=42,stratify=y)
print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)
print("\nTraining Target:", y_train.shape)
print("Testing Target:", y_test.shape)
print("\nTraining Distribution:")
print(y_train.value_counts(normalize=True) * 100)
print("\nTesting Distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
```

```python
numerical_features = [  'release_year', 'added_year', 'added_month','added_day', 'added_weekday', 
            'duration_value', 'genre_count', 'country_count', 'has_director', 'content_age'
]

categorical_features = [ 'primary_genre', 'primary_country', 'rating'
]
```

```python
numerical_transformer = Pipeline( steps=[ ('imputer', SimpleImputer(strategy='median')), ('scaler', StandardScaler()) ])

categorical_transformer = Pipeline(steps=[ ('imputer', SimpleImputer(strategy='most_frequent')),  
                                          ('encoder', OneHotEncoder(handle_unknown='ignore')) ]
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)
print("Original Training Shape:", X_train.shape)
print("Processed Training Shape:", X_train_processed.shape)
print("\nOriginal Testing Shape:", X_test.shape)
print("Processed Testing Shape:", X_test_processed.shape)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression(max_iter=1000,random_state=42)
model.fit(X_train_processed,y_train)
y_pred=model.predict(X_test_processed)
```

```python
print("Numerical Features:", numerical_features)
print("Categorical Features:", categorical_features)
```

```python

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score,classification_report
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)
lr_model.fit(X_train_processed, y_train)
y_pred_lr = lr_model.predict(X_test_processed)

print("Accuracy:", accuracy_score(y_test, y_pred_lr))
print("Precision:", precision_score(y_test, y_pred_lr, pos_label='TV Show'))
print("Recall:", recall_score(y_test, y_pred_lr, pos_label='TV Show'))
print("F1 Score:", f1_score(y_test, y_pred_lr, pos_label='TV Show'))
```

```python
lr_results = {
    'Model': 'Logistic Regression',
    'Accuracy': accuracy_score(y_test, y_pred_lr),
    'Precision': precision_score(y_test, y_pred_lr, pos_label='TV Show'),
    'Recall': recall_score(y_test, y_pred_lr, pos_label='TV Show'),
    'F1 Score': f1_score(y_test, y_pred_lr, pos_label='TV Show')
}

print(lr_results)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt_model=DecisionTreeClassifier(random_state=42,max_depth=10)
dt_model.fit(X_train_processed,y_train)
y_pred_dt=dt_model.predict(X_test_processed)
```

```python
print("Decision Tree Results")
print("Accuracy:", accuracy_score(y_test, y_pred_dt))
print("Precision:", precision_score(y_test, y_pred_dt, pos_label='TV Show'))
print("Recall:", recall_score(y_test, y_pred_dt, pos_label='TV Show'))
print("F1 Score:", f1_score(y_test, y_pred_dt, pos_label='TV Show'))
print("\nClassification Report:")
print(classification_report(y_test, y_pred_dt))
```

```python
dt_results = {
    'Model': 'Decision Tree',
    'Accuracy': accuracy_score(y_test, y_pred_dt),
    'Precision': precision_score(y_test, y_pred_dt, pos_label='TV Show'),
    'Recall': recall_score(y_test, y_pred_dt, pos_label='TV Show'),
    'F1 Score': f1_score(y_test, y_pred_dt, pos_label='TV Show')
}
print(dt_results)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)
rf_model.fit(X_train_processed, y_train)
y_pred_rf = rf_model.predict(X_test_processed)
```

```python
print("Random Forest Results")
print("Accuracy:", accuracy_score(y_test, y_pred_rf))
print("Precision:", precision_score(y_test, y_pred_rf, pos_label='TV Show'))
print("Recall:", recall_score(y_test, y_pred_rf, pos_label='TV Show'))
print("F1 Score:", f1_score(y_test, y_pred_rf, pos_label='TV Show'))
print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

```python
rf_results = {
    'Model': 'Random Forest',
    'Accuracy': accuracy_score(y_test, y_pred_rf),
    'Precision': precision_score(y_test, y_pred_rf, pos_label='TV Show'),
    'Recall': recall_score(y_test, y_pred_rf, pos_label='TV Show'),
    'F1 Score': f1_score(y_test, y_pred_rf, pos_label='TV Show')
}

print(rf_results)
```
